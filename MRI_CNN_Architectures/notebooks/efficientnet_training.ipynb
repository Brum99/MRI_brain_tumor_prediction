## Efficientnet Model Training

```python
import os
import sys

# Add parent directory to Python path to access pipeline_helpers
sys.path.append('..')

# Import data preprocessing and evaluation utility functions
from pipeline_helpers.data_preprocessing import prepare_generators
from pipeline_helpers.evaluation import (
    save_model_architecture,
    save_model_weights,
    plot_training_metrics,
    evaluate_and_save_results,
    save_confusion_matrix
)

# === Load Data ===
# Define the base path to the dataset archive
base_dir = os.path.abspath(os.path.join('..', 'archive'))

# Load training, validation, and testing data generators
train_gen, val_gen, test_gen, class_dict = prepare_generators(base_dir)

# === Load Model Architecture ===
# Import and build the EfficientNet model
from models.efficientnet_model import build_model
model = build_model()

# === Train Model ===
# Train the model on training data and validate on validation data
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=10,
    verbose=1
)

# === Define Output Directory ===
# Set output path to save model-related files and create the folder if needed
output_dir = '../outputs/efficientnet'
os.makedirs(output_dir, exist_ok=True)

# === Save Model and Metrics ===

# Save the model architecture as a PNG diagram
save_model_architecture(model, output_dir, 'efficientnet_architecture.png')

# Save model weights in both .h5 and .keras formats
save_model_weights(model, output_dir, 'efficientnet_model')

# Plot and save training metrics (accuracy, loss, precision, recall)
plot_training_metrics(history, os.path.join(output_dir, 'efficientnet_training.png'))

# Evaluate and save performance metrics on train, val, and test sets
evaluate_and_save_results(model, train_gen, val_gen, test_gen, output_dir)

# Extract class labels for plotting the confusion matrix
class_labels = list(class_dict.keys())

# Generate and save confusion matrix for test predictions
save_confusion_matrix(model, test_gen, class_labels, os.path.join(output_dir, 'efficientnet_cm.png'))
```
