## Xception Model Training

```python
import os
import sys

# Add parent directory to Python path to access pipeline_helpers
sys.path.append('..')

# Import custom data preprocessing and evaluation utilities
from pipeline_helpers.data_preprocessing import prepare_generators
from pipeline_helpers.evaluation import (
    save_model_architecture,
    save_model_weights,
    plot_training_metrics,
    evaluate_and_save_results,
    save_confusion_matrix
)

# === Load Data ===
# Define the base path to the dataset archive
base_dir = os.path.abspath(os.path.join('..', 'archive'))

# Load the data generators for training, validation, and testing
train_gen, val_gen, test_gen, class_dict = prepare_generators(base_dir)

# === Load Model Architecture ===
# Import the Xception model architecture
from models.xception_model import build_xception_model

# Build the Xception model
model = build_xception_model(input_shape=(224,224,3), num_classes=4)

# === Train Model ===
# Train the model using the training and validation generators
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=10,
    verbose=1
)

# === Define Output Directory ===
# Path to save model outputs, plots, and metrics
output_dir = '../outputs/xception'
os.makedirs(output_dir, exist_ok=True)

# === Save Model and Metrics ===

# Save model architecture as a PNG image
save_model_architecture(model, output_dir, 'xception_architecture.png')

# Save model weights in .h5 and .keras formats
save_model_weights(model, output_dir, 'xception_model')

# Plot and save training accuracy, loss, precision, and recall curves
plot_training_metrics(history, os.path.join(output_dir, 'xception_training.png'))

# Evaluate the model on training, validation, and test sets and save results
evaluate_and_save_results(model, train_gen, val_gen, test_gen, output_dir)

# Extract class labels from the class dictionary
class_labels = list(class_dict.keys())

# Generate and save the confusion matrix plot
save_confusion_matrix(model, test_gen, class_labels, os.path.join(output_dir, 'xception_cm.png'))
```
