## Xception Model Training

```
import os
import sys
sys.path.append('..')
from pipeline_helpers.data_preprocessing import prepare_generators
from pipeline_helpers.evaluation import (save_model_architecture, save_model_weights, plot_training_metrics, evaluate_and_save_results, save_confusion_matrix)
```

```
# Update this path to your dataset location
base_dir = os.path.abspath(os.path.join('..', 'archive'))
train_gen, val_gen, test_gen, class_dict = prepare_generators(base_dir)
```

```
from models.xception_model import build_model
model = build_model()
```

```
history = model.fit(train_gen, validation_data=val_gen, epochs=10, verbose=1)
```

```
output_dir = '../outputs/models'
os.makedirs(output_dir, exist_ok=True)

save_model_architecture(model, output_dir, f'xception_architecture.png')
save_model_weights(model, output_dir, f'xception_model')
plot_training_metrics(history, os.path.join(output_dir, f'xception_training.png'))
evaluate_and_save_results(model, train_gen, val_gen, test_gen, output_dir)
class_labels = list(class_dict.keys())
save_confusion_matrix(model, test_gen, class_labels, os.path.join(output_dir, f'xception_cm.png'))
```
